# 3. Estructuras de control: `if`, `elif` y `else`

Las estructuras condicionales permiten que el programa tome decisiones.

- `if` ejecuta un bloque si una condición es verdadera.
- `elif` permite evaluar otra condición si la anterior fue falsa.
- `else` se ejecuta cuando ninguna condición anterior se cumplió.

Los bloques se reconocen por su **indentación**. Después de cada condición se escriben dos puntos (`:`), y las instrucciones que dependen de ella deben quedar desplazadas hacia la derecha.

## Machete de comparaciones

| Operador | Significado |
|---|---|
| `==` | Igual a |
| `!=` | Distinto de |
| `>` | Mayor que |
| `<` | Menor que |
| `>=` | Mayor o igual que |
| `<=` | Menor o igual que |

## Ejemplo ejecutable: clasificar una temperatura

Este ejemplo muestra una estructura con tres caminos posibles.

````python
temperatura = float(input("Ingresá la temperatura: "))

if temperatura < 18:
    print("La temperatura es baja.")
elif temperatura <= 25:
    print("La temperatura es moderada.")
else:
    print("La temperatura es alta.")
````

# 4. Estructuras repetitivas: `while`

Un ciclo `while` repite un bloque de instrucciones **mientras** una condición sea verdadera.

Es importante que dentro del ciclo ocurra algún cambio que permita terminarlo. Si la condición nunca deja de ser verdadera, se produce un ciclo infinito. Si eso sucede en Jupyter, podés detener la ejecución con el botón de interrupción del cuaderno.

## Ejemplo ejecutable: cuenta regresiva

Seguí el valor de `contador` en cada repetición y observá por qué el ciclo finalmente termina.

````python
contador = 3

while contador > 0:
    print(contador)
    contador = contador - 1

print("Fin de la cuenta regresiva.")
````

## Ejemplo ejecutable: repetir hasta recibir una palabra

En este caso no conocemos de antemano la cantidad de repeticiones. El ciclo termina cuando el usuario escribe una palabra determinada.

````python
respuesta = ""

while respuesta != "salir":
    respuesta = input("Escribí algo o ingresá 'salir' para terminar: ")
    print("Ingresaste:", respuesta)

print("Programa finalizado.")
````

# 5. Estructuras repetitivas: `for`

Un ciclo `for` se utiliza para recorrer elementos o repetir instrucciones una cantidad determinada de veces.

Puede recorrer directamente una lista, un texto u otra colección. También puede trabajar junto con `range()` para generar una secuencia de números.

La forma `range(inicio, fin)` incluye el valor inicial, pero **no incluye** el valor final. Por ejemplo, `range(1, 4)` genera `1`, `2` y `3`.

## Ejemplo ejecutable: recorrer una lista

````python
herramientas = ["destornillador", "pinza", "multímetro"]

for herramienta in herramientas:
    print("Herramienta:", herramienta)
````

## Ejemplo ejecutable: repetir con `range()`

Modificá el inicio y el final para observar cómo cambia la secuencia.

````python
for numero in range(1, 4):
    print("Repetición", numero)
````

## Ejercicio 4: escalera de asteriscos

Creá un programa que pida la cantidad de niveles de una escalera.

- La cantidad debe ser mayor que `0` y menor que `10`.
- Si el valor no es válido, el programa debe volver a pedirlo.
- Cuando el valor sea válido, usá `for` para mostrar la escalera.

Para 6 niveles, el formato esperado es:

```text
*
**
***
****
*****
******
```

> En Python, un texto puede repetirse al multiplicarlo por un número entero.

````python
niveles=int(input("Dime la cantidad de niveles de la escalera(P.D: Debe ser mayor a 0 y menor a 10)"))
if 0<niveles and niveles<10:
    for niveles in range (1, niveles+1):
        print("*" * niveles)
    
````
